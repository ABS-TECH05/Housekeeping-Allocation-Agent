```python
housekeeping_df = df.copy()
```

```python
keep_cols = [
    'hotel_id',
    'hotel_name',
    'hotel_rank',
    'hotel_no_reviews',
    'rating',
    'review_date',
    'review_content',
    'excellent',
    'very_good',
    'average',
    'poor',
    'terrible',
    'families',
    'couples',
    'solo',
    'business',
    'friends'
]

housekeeping_df = housekeeping_df[keep_cols]

housekeeping_df.head()
```

```python
housekeeping_df.isnull().sum()
```

```python
housekeeping_df = housekeeping_df.drop_duplicates()

print(housekeeping_df.shape)
```

```python
housekeeping_df.isnull().sum()
```

```python
housekeeping_df = housekeeping_df.dropna(
    subset=["review_content"]
)

housekeeping_df = housekeeping_df.fillna(0)

print(housekeeping_df.shape)
```

```python
housekeeping_df.isnull().sum().sum()
```

```python
housekeeping_df["room_id"] = np.random.randint(
    100,
    999,
    size=len(housekeeping_df)
)
```

```python
housekeeping_df["room_id"] = range(
    1000,
    1000 + len(housekeeping_df)
)
```

```python
complaint_keywords = [
    "dirty",
    "unclean",
    "dust",
    "stain",
    "smell",
    "housekeeping",
    "cleaning",
    "filthy",
    "bathroom",
    "toilet"
]

housekeeping_df["complaint_flag"] = (
    housekeeping_df["review_content"]
    .astype(str)
    .str.lower()
    .apply(
        lambda text: int(
            any(word in text for word in complaint_keywords)
        )
    )
)

print("Done")
```

```python
housekeeping_df["complaint_flag"].value_counts()
```

```python
def assign_priority(rating):

    if rating <= 2:
        return "High"

    elif rating == 3:
        return "Medium"

    else:
        return "Low"


housekeeping_df["housekeeping_priority"] = (
    housekeeping_df["rating"]
    .apply(assign_priority)
)

housekeeping_df["housekeeping_priority"].value_counts()
```

```python
housekeeping_df["cleaning_demand_score"] = (
    housekeeping_df["poor"]
    + housekeeping_df["terrible"]
    - (0.5 * housekeeping_df["excellent"])
)

housekeeping_df["cleaning_demand_score"].describe()
```

```python
guest_cols = [
    "families",
    "couples",
    "solo",
    "business",
    "friends"
]

housekeeping_df["dominant_guest_type"] = (
    housekeeping_df[guest_cols]
    .idxmax(axis=1)
)

housekeeping_df["dominant_guest_type"].value_counts()
```

```python
np.random.seed(42)

housekeeping_df["room_status"] = np.where(
    housekeeping_df["complaint_flag"] == 1,
    "Dirty",
    "Clean"
)

housekeeping_df["checkout_hour"] = np.random.randint(
    6,
    12,
    size=len(housekeeping_df)
)

housekeeping_df["estimated_cleaning_time"] = np.where(
    housekeeping_df["room_status"] == "Dirty",
    np.random.randint(30, 60, size=len(housekeeping_df)),
    np.random.randint(10, 20, size=len(housekeeping_df))
)
```
